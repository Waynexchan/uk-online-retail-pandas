# Q2: Top Products and Customer Concentration Analysis

### Business Question
1. Which products contribute most to revenue?
2. Is revenue concentrated among a small group of customers?

### Objective
Evaluate product-level revenue distribution and assess customer concentration risk.

```python
import pandas as pd
df = pd.read_csv('../data/processed/retail_cleaned.csv')
df.head()
```

```python
## Determine the top 10 products by total sales
total_sales_by_stockcode = df.groupby('StockCode')['TotalSales'].sum().sort_values(ascending=False).head(10)
print(total_sales_by_stockcode)
```

```python
total_revenue = df['TotalSales'].sum()
print(total_revenue)
```

```python
top_products_share = (total_sales_by_stockcode / total_revenue) * 100
print(top_products_share)
```

```python
top_products_share.sum()
```

```python
## Exclude non-product items and recalculate the top 10 products by total sales
exclude_codes = ['POST', 'M', 'DOT']
df_filtered = df[~df['StockCode'].isin(exclude_codes)]
total_sales_by_stockcode_filtered = df_filtered.groupby('StockCode')['TotalSales'].sum().sort_values(ascending=False).head(10)
print(total_sales_by_stockcode_filtered)
```

```python
total_filtered_revenue = df_filtered['TotalSales'].sum()
print(total_filtered_revenue)
```

```python
top_filtered_products_share = (total_sales_by_stockcode_filtered / total_filtered_revenue) * 100
print(top_filtered_products_share)
```

```python
top_filtered_products_share.sum()
```

## Top Products by Revenue

### Method
Group by StockCode and calculate total revenue. Exclude non-product codes such as POST, DOT, and M.

### Insight
The top 10 products account for approximately 8% of total product revenue, indicating a diversified product portfolio.


```python
df.head()
```

```python
## Analyze top customers by total sales
df_customers = df[df['Customer_ID'].notna()]
```

```python
top_customers = df_customers.groupby('Customer_ID')['TotalSales'].sum().sort_values(ascending=False).head(10)
total_customer_revenue = df_customers['TotalSales'].sum()
top_customers_share = (top_customers/ total_customer_revenue) * 100
print(top_customers_share)
```

```python
top_customers_share.sum()
```

## Customer Revenue Concentration

### Method
Group by Customer_ID and calculate total revenue per customer. Compute revenue share for top customers.

### Insight
The top 10 customers contribute approximately 16% of total identified customer revenue, suggesting limited dependency risk.


```python
## Pareto analysis of customer revenue concentration
customer_revenue = df_customers.groupby('Customer_ID')['TotalSales'].sum()
customer_revenue_sorted = customer_revenue.sort_values(ascending=False)
cumulative_revenue = customer_revenue_sorted.cumsum()
cumulative_percentage = (cumulative_revenue / total_customer_revenue) * 100
print(cumulative_percentage.head(20))
```

```python
total_customers = len(customer_revenue_sorted)
print(f"Total customers: {total_customers}")
```

```python
top_20_percent_customers = int(total_customers * 0.2)
```

```python
cumulative_percentage.iloc[top_20_percent_customers]
```

## Pareto Analysis Result
Pareto analysis reveals that the top 20% of customers contribute approximately 77% of total revenue. This indicates a structurally concentrated customer base consistent with a typical 80/20 distribution.
