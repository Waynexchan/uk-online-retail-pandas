```python
import pandas as pd
df = pd.read_csv('../data/processed/retail_cleaned.csv')
df.head()
```

```python
total_sales_by_stockcode = df.groupby('StockCode')['TotalSales'].sum().sort_values(ascending=False).head(10)
print(total_sales_by_stockcode)
```

```python
total_revenue = df['TotalSales'].sum()
print(total_revenue)
```

```python
top_products_share = (total_sales_by_stockcode / total_revenue) * 100
print(top_products_share)
```

```python
top_products_share.sum()
```

```python
exclude_codes = ['POST', 'M', 'DOT']
df_filtered = df[~df['StockCode'].isin(exclude_codes)]
total_sales_by_stockcode_filtered = df_filtered.groupby('StockCode')['TotalSales'].sum().sort_values(ascending=False).head(10)
print(total_sales_by_stockcode_filtered)
```

```python
total_filtered_revenue = df_filtered['TotalSales'].sum()
print(total_filtered_revenue)
```

```python
top_filtered_products_share = (total_sales_by_stockcode_filtered / total_filtered_revenue) * 100
print(top_filtered_products_share)
```

```python
top_filtered_products_share.sum()
```
