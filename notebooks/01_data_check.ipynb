```python
missing_pct = df.isna().sum()*100/len(df)
missing_pct[missing_pct > 0]
```

```python
quantity_over_zero = df[df['Quantity'] > 0]
quantity_over_zero.shape

```

```python
price_over_zero = quantity_over_zero[quantity_over_zero['Price'] > 0]
price_over_zero.shape
```

```python
remove_start_c = price_over_zero[~price_over_zero['Invoice'].astype(str).str.startswith('C')]
remove_start_c.shape
```

```python
remove_start_c['TotalSales'] = remove_start_c['Quantity'] * remove_start_c['Price']
remove_start_c.head(5)
```
