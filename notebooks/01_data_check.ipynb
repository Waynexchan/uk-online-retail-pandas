```python
import pandas as pd
```

```python
df1= pd.read_excel('../data/raw/online_retail_II.xlsx', sheet_name='Year 2009-2010')
df2=pd.read_excel('../data/raw/online_retail_II.xlsx', sheet_name='Year 2010-2011')
df_all = pd.concat([df1,df2], ignore_index=True)
```

```python
df_all.columns = df_all.columns.str.strip()
df_all.columns = df_all.columns.str.replace(' ', '_')
df_all.columns
```

```python
df_all = df_all[df_all['Price'] > 0]
```

```python
df_all['TotalSales'] = df_all['Quantity'] * df_all['Price']
df_all.head(5)
```

```python
df_all['Customer_ID'] = df_all['Customer_ID'].astype('Int64')
df_all['StockCode'] = df_all['StockCode'].astype(str)
```

```python
df_all.to_csv('../data/processed/retail_with_returns.csv', index=False)
```
